```python
import uproot
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import trapezoid  # Updated to trapezoid

import mplhep as mpl

mpl.style.use('ATLAS')
layers = [1, 3, 5, 7, 9, 11, 13, 15]

files = []
for i in range(1, 8 + 1):
    filename = "Squarecoordinates_thickness" + str(i) + ".root"
    files.append(filename)

# Lists for efficiencies and errors
efficiencys = []
efficiencys_err = []
reweight_efficiencys = []
reweight_efficiencys_err = []

# Lists to collect positions and weights across all files for verification plots
all_posz = []
all_wz = []
all_rho = []
all_wpt = []
all_phi = []
all_wpp = []

# Function to load .dat files (assumes two columns: upper_edge weight)
def load_bias_dat(filename):
    data = np.loadtxt(filename)
    uppers = data[:, 0]
    weights = data[:, 1]
    # Sort if not already (for interp and integration)
    sort_idx = np.argsort(uppers)
    return uppers[sort_idx], weights[sort_idx]

# Load bias data once outside loop
uppers_z, weights_z = load_bias_dat('biasz.dat')
uppers_pt, weights_pt = load_bias_dat('biaspt.dat')
uppers_pp, weights_pp = load_bias_dat('biaspp.dat')  # Assuming biasspp.dat is for phi (biaspp)

for file_ in files:
    # Load the original ROOT file
    with uproot.open(file_) as file:
        # Extract relevant NTuples
        edep_tree = file["EdepCounter"]
        truth_gamma_tree = file["TruthGamma"]
        truth_ps_tree = file["TruthPs"]
        truth_positron_tree = file["TruthPositron"]
        # Convert to DataFrames
        edep_df = edep_tree.arrays(library="pd")
        truth_gamma_df = truth_gamma_tree.arrays(library="pd")
        truth_ps_df = truth_ps_tree.arrays(library="pd")
        truth_positron_df = truth_positron_tree.arrays(library="pd")

    # Process TruthGamma to split into oPs, pPs, others (unchanged)
    o_ps_ids = truth_ps_df[truth_ps_df['Type'] == 'o-Ps']['TrackID'].unique()
    p_ps_ids = truth_ps_df[truth_ps_df['Type'] == 'p-Ps']['TrackID'].unique()
    truth_gamma_ops_df = truth_gamma_df[truth_gamma_df['ParentID'].isin(o_ps_ids)]
    truth_gamma_pps_df = truth_gamma_df[truth_gamma_df['ParentID'].isin(p_ps_ids)]
    truth_gamma_others_df = truth_gamma_df[~(truth_gamma_df['ParentID'].isin(o_ps_ids) | truth_gamma_df['ParentID'].isin(p_ps_ids))]

    # EdepCounter: Group by EventID, DetectorName, FirstTime_ns, sum Edep_MeV
    edep_grouped = edep_df.groupby(['EventID', 'DetectorName', 'FirstTime_ns'])['Edep_MeV'].sum().reset_index()

    # Rebuild edep_df (unchanged, but simplified)
    edep_df = pd.DataFrame({
        "EventID": edep_grouped['EventID'].values,
        "DetectorName": edep_grouped['DetectorName'].values,
        "Edep_MeV": edep_grouped['Edep_MeV'].values,
        "FirstTime_ns": edep_grouped['FirstTime_ns'].values
    })

    # Events with exactly 3 hits
    common_events = edep_df.groupby("EventID").filter(lambda x: len(x) == 3)

    # Compute Max_Edep, Total_Edep, Remaining_Sum per event
    common_events['Max_Edep'] = common_events.groupby('EventID')['Edep_MeV'].transform('max')
    common_events['Total_Edep'] = common_events.groupby('EventID')['Edep_MeV'].transform('sum')
    common_events["Remaining_Sum"] = common_events['Total_Edep'] - common_events['Max_Edep']

    # Get unique EventIDs for all events with 3 hits
    all_3hit_events = common_events['EventID'].unique()

    # Get Total_Edep per event (unique)
    event_edep = common_events.groupby('EventID')['Total_Edep'].first()

    # Detection condition: Total_Edep between 1 and 1.3
    detected_mask = event_edep.between(1, 1.3)

    # Number of detected events
    num_detected = detected_mask.sum()

    # Original efficiency and error (fixed to use unique events)
    total_events = 100000  # Assumed fixed; replace with len(truth_ps_df['EventID'].unique()) if varies
    eff = num_detected / total_events
    eff_err = np.sqrt(num_detected) / total_events  # Approximate Poisson error
    print(f"File={file_}")
    print(f"o-Ps, detection efficiency={eff} ± {eff_err}")
    efficiencys.append(eff)
    efficiencys_err.append(eff_err)

    # Extract o-Ps positions (assuming one o-Ps per event; if multiple, aggregate or filter)
    ops_df = truth_ps_df[truth_ps_df['Type'] == 'o-Ps'][['EventID', 'PosX_mm', 'PosY_mm', 'PosZ_mm']].copy()

    # Compute rho and phi
    ops_df['rho_mm'] = np.sqrt(ops_df['PosX_mm']**2 + ops_df['PosY_mm']**2)
    ops_df['phi'] = np.arctan2(ops_df['PosY_mm'], ops_df['PosX_mm'])
    ops_df['phi'] = (ops_df['phi'] + 2 * np.pi) % (2 * np.pi)  # Shift to [0, 2π) assuming .dat range

    # Assign detected (align with ops_df index)
    ops_df = ops_df.set_index('EventID')
    ops_df['detected'] = detected_mask.reindex(ops_df.index, fill_value=False)
    ops_df = ops_df.reset_index()

    # Interpolate weights (linear, 0 outside range)
    ops_df['w_z'] = np.interp(ops_df['PosZ_mm'], uppers_z, weights_z, left=0, right=0)
    ops_df['w_pt'] = np.interp(ops_df['rho_mm'], uppers_pt, weights_pt, left=0, right=0)
    ops_df['w_pp'] = np.interp(ops_df['phi'], uppers_pp, weights_pp, left=0, right=0)

    # Jacobian correction for rho (assuming circular cross-section, original pdf ∝ rho)
    ops_df['w_pt'] /= (ops_df['rho_mm'] + 1e-6)  # Avoid division by zero

    ops_df['weight'] = ops_df['w_z'] * ops_df['w_pt'] * ops_df['w_pp']

    # Total weight over all o-Ps events
    total_weight = ops_df['weight'].sum()

    # Reweighted detected weight and error
    detected_weights = ops_df[ops_df['detected']]['weight']
    reweight_detected_weight = detected_weights.sum()
    reweight_detected_err = np.sqrt((detected_weights**2).sum())  # Approximate weighted Poisson error

    # Reweighted efficiency and error
    reweight_eff = reweight_detected_weight / total_weight if total_weight > 0 else 0
    reweight_eff_err = reweight_detected_err / total_weight if total_weight > 0 else 0
    print(f"Reweighted o-Ps, detection efficiency={reweight_eff} ± {reweight_eff_err}")
    reweight_efficiencys.append(reweight_eff)
    reweight_efficiencys_err.append(reweight_eff_err)

    # Collect for verification plots
    all_posz.extend(ops_df['PosZ_mm'])
    all_wz.extend(ops_df['w_z'])
    all_rho.extend(ops_df['rho_mm'])
    all_wpt.extend(ops_df['w_pt'])
    all_phi.extend(ops_df['phi'])
    all_wpp.extend(ops_df['w_pp'])

# Plot efficiencies with error bars
plt.figure()
plt.errorbar(layers, efficiencys, yerr=efficiencys_err, capsize=3, marker='o', linestyle='-', label="Uniform (original)")
plt.errorbar(layers, reweight_efficiencys, yerr=reweight_efficiencys_err, capsize=3, marker='o', linestyle='-', label="Reweighted")
plt.xlabel("Layer")
plt.ylabel("Efficiency")
plt.legend()
plt.title("Detection Efficiency vs. Layer")
plt.show()

# Verification plots (density histograms: uniform vs. reweighted, overlaid with normalized target pdf)
def plot_verification(values, weights, uppers, target_weights, xlabel, title,x_lim=None):
    fig, ax = plt.subplots()
    ax.hist(values, bins=100, density=True, label='Uniform (actual)')
    ax.hist(values, bins=100, weights=weights, density=True, histtype='step', label='Reweighted')
    # Normalize target pdf
    fine_uppers = np.linspace(min(uppers), max(uppers), 1000)
    fine_weights = np.interp(fine_uppers, uppers, target_weights)
    norm = trapezoid(fine_weights, fine_uppers)
    if norm > 0:
        ax.plot(fine_uppers, fine_weights / norm, label='Target pdf (linear interp from .dat)')
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Density')
    ax.set_title(title)
    if x_lim!=None:
        ax.set_xlim(0,9)
    ax.legend()
    plt.show()

plot_verification(all_posz, all_wz, uppers_z, weights_z, 'PosZ_mm', 'Verification for Z Bias')
plot_verification(all_phi, all_wpp, uppers_pp, weights_pp, 'phi (radians)', 'Verification for Phi (pp) Bias')
plot_verification(all_rho, all_wpt, uppers_pt, weights_pt, 'rho_mm', 'Verification for Transverse (pt/rho) Bias',x_lim="tr")
```
